```python
# Ex. 5
# 1.
import math
h = 0.001
a = 0.5

u0 = math.exp(0)
u1 = math.exp(h * a)
u2 = math.exp(2 * h * a)
u = [u0, u1, u2]
for n in range(1000-3):
    #because we already have u0 u1 & u2 so we need only 997 elements 
    un = u[n]
    un1 = u[n+1]
    un2 = u[n+2]

    un3 = un2 + h * a * ( (23/12) * un2 - (4/3) * un1 + (5/12) * un)

    u.append(un3)
len(u) 
```

```python
# Ex. 6
A = {1, 2, 3, 4}
B = {2, 3, 4, 5}


def symmetric_difference(A,B):
    A_difference_B = A.difference(B)
    B_difference_A = B.difference(A)

    union = A_difference_B.union(B_difference_A)
    return union

A.symmetric_difference(B)
```

```python
# Ex. 8
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(f"Stes of work: A = {A}, B = {B}")

print(f"A.update(B) returns: {A.update(B)}")
print(f"Stes after update(): A = {A}, B = {B}")

print(f"A.intersection_update(B) returns: {A.intersection_update(B)}")
print(f"Stes after intersection_update(): A = {A}, B = {B}")

A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(f"Stes before work: A = {A}, B = {B}")

print(f"A.intersection(B) returns: {A.intersection(B)}")
print(f"Stes after updateintersection(): A = {A}, B = {B}")

print(f"A.intersection_update(B) returns: {A.intersection_update(B)}")
print(f"Stes after intersection_update(): A = {A}, B = {B}")
```

```python
# with update() all elements are in A now 
# with intersection_update() → keep ONLY elements that are in BOTH A and B 
# intersection() shows what is both sets but it didn't change A 
```
